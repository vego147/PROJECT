```python

from pathlib import Path
from dotenv import load_dotenv, dotenv_values
import os

backend_dir = Path.cwd().parent  # tests/ -> backend/
env_path = backend_dir / ".env"

print("Looking for:", env_path)
print("File exists:", env_path.is_file())

if env_path.is_file():
    values = dotenv_values(env_path)

    print("Key exists in .env:", bool(values.get("GROQ_API_KEY")))
    print("Model in .env:", values.get("GROQ_MODEL", "(not set)"))

    load_dotenv(env_path, override=True)

    print("Key loaded into environment:", bool(os.getenv("GROQ_API_KEY")))
else:
    print("ERROR: No .env file found in backend/")
```
